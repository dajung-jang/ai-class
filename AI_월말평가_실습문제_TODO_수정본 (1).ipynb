## TODO 1-2. IQR 기반 컬럼별 이상치 개수

타겟 `target`을 제외한 수치형 특성에 대해 다음을 계산하세요.

- `Q1`: 하위 25% 지점
- `Q3`: 상위 75% 지점
- `IQR = Q3 - Q1`
- 하한: `Q1 - 1.5 * IQR`
- 상한: `Q3 + 1.5 * IQR`
- 각 컬럼에서 경계를 벗어난 값의 개수를 `outlier_counts`에 저장하세요.


```python
# TODO 1-2
# feature_cols = df.columns.drop("target")

# quatile(q) = 데이터를 크기순으로 정렬했을때 하위 q% (0~1 사이) 지점의 값 구해주는 함수
Q1 = df['flavanoids'].quantile(0.25)    #하위 25% 지점 = 1사분위수 (작은 순서대로 쭉 나열했을때 앞에서 25% 지점에 있는 값)
Q3 = df['flavanoids'].quantile(0.75)    #하위 75% 지점 = 3사분위수
IQR = Q3 - Q1  # 데이터 중간 50% 가 얼마나 넓게 퍼져있는지 나타내는 값 / IQR 이 크면 중간값들도 넓게 퍼져있다는 뜻

# 이상치를 판단하는 경계선 만드는 공식
# Q1 보다 IQR 의 1.5배 만큼 더 작은값, Q3보다 IQR의 1.5배 만큼 더 큰값을 각각 하한/상한으로 잡음
# 이 범위를 벗어나는 값을 "극단적으로 튀는 값"으로 취급
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# flavanoids 값이 정상 범위 (lower~upper) 안에 있는 행들만 남음
df_no_outlier = df[(df['flavanoids'] >= lower) & (df['flavanoids'] <= upper)]
display(df_no_outlier)

# 이상치 안에 있는 행들만 남음
df_outlier = df[(df['flavanoids'] < lower) | (df['flavanoids'] > upper)]
display(df_outlier)
```

## TODO 1-3. 상관계수 출력

1. 전체 수치형 컬럼의 상관행렬을 `corr_matrix`에 저장하세요.
2. `flavanoids`와 `total_phenols`의 피어슨 상관계수를 `corr_value`에 저장하고 출력하세요.


```python
# TODO 1-3

# axios = 1 => 행이 아니라 열을 지운다는 뜻(컬럼을 지움) / df 에서 flavanoids 컬럼을 지움(flavanoids 기준으로 상관관계 계산해야해서)
# .corrwith(df['flavanoids']): 방금 만든 (flavanoids 제외한) DataFrame의 각 컬럼 하나하나와 df['flavanoids'] 하나의 상관계수를 계산해줘.
corr_with_flav = df.drop('flavanoids', axis=1).corrwith(df['flavanoids'])

# abs() : 모든 상관계수 값을 절댓값으로 바꿈(양, 음 상관없이 관계 세기만 비교하려고)
# idxmax() : 절대값이 가장 큰 값의 인덱스(변수명)를 반환 (max()였다면 값 자체가 나왔을텐데, idx 가 붙어서 그 값의 위치/이름 을 가져오는 것)
top_corr_feature = corr_with_flav.abs().idxmax()
# 위에서 절대값 기준으로 변수명만 찾았으니 그 변수의 원래 부호 있는 상관계수 값을 다시 꺼내는 것
# (양수인지 음수인지 알아야 "같이 늘어나는 관계"인지 "반대로 움직이는 관계"인지 알 수 있으니까)
top_corr_value = corr_with_flav[top_corr_feature]


print(f"flavanoids와 가장 강한 상관관계를 가진 특성 : {top_corr_feature}")
print(f"상관계수: {top_corr_value:.4f}")
```
